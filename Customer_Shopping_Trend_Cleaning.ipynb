```python
#changing column names into snake case for better usability
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df= df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#creating a new column to group customers as per their age
labels= ['Young', 'Adult', 'Middle-Aged', 'Senior']
df['age_group']= pd.qcut(df['age'], 4 , labels=labels)
```

```python
df[['age', 'age_group']].head(10)
```

```python
df['frequency_of_purchases'].unique()
```

```python
df['frequency_of_purchases'] = df['frequency_of_purchases'].replace({'Fortnightly' : 'Bi-Weekly', 'Every 3 Months' : 'Quarterly'})
```

```python
df['frequency_of_purchases'].unique()
```

```python
#creating a new column to judge purchase frequency days
frquency_mapping={
    'Weekly' : 7,
    'Annually' : 365,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Monthly' : 30
}

df['purchase_frequency_days']= df['frequency_of_purchases'].map(frquency_mapping)
```

```python
df[['frequency_of_purchases', 'purchase_frequency_days']].head(10)
```

```python
# checking the 2 columns 'discount_applied' and 'promo_code_used', the logic here is whenever a promo code will be used the discount will be there but a discount may be there in some cases where no promo code is used. so we will check if the columns are redundant and if we can remove any
df[['discount_applied', 'promo_code_used']].head(15)
```

```python
#checking if the values in both the column same for all the rows
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
df= df.drop('promo_code_used', axis=1) # dropping the column 'promo_code_applied' as it is redundant in this case
```
